```python
# Task 18: "Students Marks Analyzer"
student_name = input("Enter student name: ")

subjects = 5
marks_per_subject = 100

maximum_marks = subjects * marks_per_subject

marks = []

english_marks = int(input("Enter english marks: "))
urdu_marks = int(input("Enter urdu marks: "))
islamiat_marks = int(input("Enter islamiat marks: "))
maths_marks = int(input("Enter maths marks: "))
science_marks = int(input("Enter science marks: "))

marks.append(english_marks)
marks.append(urdu_marks)
marks.append(islamiat_marks)
marks.append(maths_marks)
marks.append(science_marks)

total_marks = 0

for mark in marks:
    total_marks = total_marks + mark

def calculate_percentage(total):
    percentage = (total / maximum_marks) * 100
    return percentage

percentage = calculate_percentage(total_marks)

print("student name:", student_name)

if percentage >= 50:
    result = "pass"
else:
    result = "fail"

highest = max(marks)
lowest = min(marks)
number_of_subjects = len(marks)

student_record = {
    "name": student_name,
    "marks": marks,
    "total": total_marks,
    "percentage": percentage,
    "result": result,
    "highest": highest,
    "lowest": lowest
}

print("Marks:", marks)
print("Total Marks:", total_marks)
print("Percentage:", percentage)
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("Number of Subjects:", number_of_subjects)
print("Student Record:", student_record)
```
