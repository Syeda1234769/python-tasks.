```python
#Task 11 Persnol Expense Tracker
expenses = []
total_expenses = int(input("Enter the total expenses:  "))
for i in range(total_expenses):

    expense_name = input("Enter expense name: ")
    category = input("Enter category:  ").title()
    amount = int(input("Enter amount: "))

    expense = {
     "name": expense_name,
     "category": category,
     "amount": amount
    }
    expenses.append(expense)

total_spending = 0

for expense in expenses:
    total_spending += expense["amount"]

print("Total Spending:", total_spending)


food_total = 0
transport_total = 0
shopping_total = 0
bills_total = 0
other_total = 0

for expense in expenses:

    if expense["category"] == "Food":
        food_total += expense["amount"]

    elif expense["category"] == "Transport":
        transport_total += expense["amount"]

    elif expense["category"] == "Shopping":
        shopping_total += expense["amount"]

    elif expense["category"] == "Bills":
        bills_total += expense["amount"]

    else:
        other_total += expense["amount"]

print("Food:", food_total)
print("Transport:", transport_total)
print("Shopping:", shopping_total)
print("Bills:", bills_total)
print("Other:", other_total)

highest_expense = expenses[0]

for expense in expenses:

    if expense["amount"] > highest_expense["amount"]:
        highest_expense = expense

print("Highest Expense:")
print("Name:", highest_expense["name"])
print("Category:", highest_expense["category"])
print("Amount:", highest_expense["amount"])

budget = int(input("Enter your budget: "))

if total_spending > budget:
    print("WARNING: Budget Exceeded!")
    print("Exceeded Amount:", total_spending - budget)
else:
    print("Budget is within limit.")
    print("Remaining Budget:", budget - total_spending)

def final_report():
    print("========== FINAL REPORT ==========")
    print("Total Spending:", total_spending)

    print("Food:", food_total)
    print("Transport:", transport_total)
    print("Shopping:", shopping_total)
    print("Bills:", bills_total)
    print("Other:", other_total)

    print("Highest Expense:", highest_expense["name"])
    print("Highest Amount:", highest_expense["amount"])
final_report()
```

```python

# Task 12: Store Inventory Manager

products = {}

def add_product():
    name = input("Enter product name: ")
    price = float(input("Enter product price: "))
    quantity = int(input("Enter product quantity: "))

    products[name] = {"price": price, "quantity": quantity}
    print(f"{name} has been added to the inventory.")


def sell_product():
    product_name = input("Enter product name: ")

    if product_name in products:
        quantity_sold = int(input("Enter quantity sold: "))

        if quantity_sold <= products[product_name]["quantity"]:
            products[product_name]["quantity"] -= quantity_sold
            print(f"{quantity_sold} units of {product_name} have been sold.")
        else:
            print("Insufficient quantity in inventory.")
    else:
        print("Product not found in inventory.")


def restock_product():
    product_name = input("Enter product name: ")

    if product_name in products:
        quantity_restocked = int(input("Enter quantity restocked: "))
        products[product_name]["quantity"] += quantity_restocked
        print(f"{quantity_restocked} units of {product_name} have been restocked.")
    else:
        print("Product not found in inventory.")


def search_product():
    product_name = input("Enter product name: ")

    if product_name in products:
        print(f"Product: {product_name}")
        print(f"Price: {products[product_name]['price']}")
        print(f"Quantity: {products[product_name]['quantity']}")
    else:
        print("Product not found in inventory.")


def display_inventory():
    print("========== INVENTORY REPORT ==========")
    total_inventory_value = 0

    for product_name, product_info in products.items():
        print(f"Product: {product_name}")
        print(f"Price: {product_info['price']}")
        print(f"Quantity: {product_info['quantity']}")

        if product_info["quantity"] < 5:
            print("Low Stock")
        product_value = product_info["price"] * product_info["quantity"]
        total_inventory_value += product_value
        print(f"Value: {product_value}")
        print()
    print(f"Total Inventory Value: {total_inventory_value}")
    print("==================================")


while True:
    print("\n======Store Inventory Manager Menu======:")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")
    choice = int(input("Enter your choice   "))
    if choice == "1":
       print("Add product")
       add_product()
    elif choice == "2":
       print("Sell product")
       sell_product()
    elif choice == "3":
        print("Restock product")
        restock_product()
    elif choice == "4":
        print("Search product")
        search_product()
    elif choice == "5":
        print("Display inventory")
        display_inventory()
    elif choice == "6":
        print("Exit")
        break

    else:
        print("Invalid choice. Please try again.")
```

```python
#Task 13: Library Book Management System
books = []
def add_book():
    book_id = input("Enter book ID: ")
    book_title = input("Enter book title: ")
    book_author = input("Enter author: ")
    for book in books:
        if book["id"] == book_id:
           print("Book ID already exists.")
           return
    book = {
        "id": book_id,
        "title": book_title,
        "author": book_author,
        "availability" : True
    }
    books.append(book)
    print(f"Book '{book_title}' has been added to the library.")
def display_books():
    print("========== LIBRARY BOOKS ==========")
    for book in books:
        print(f"ID: {book['id']}")
        print(f"Title: {book['title']}")
        print(f"Author: {book['author']}")
        print(f"Availability: {'Available' if book['availability'] else 'Not Available'}")
        print()
        print("==================================")
def search_book():
    book_title = input("Enter book title: ")
    for book in books:
        if book["title"] == book_title:
           print(f"ID: {book['id']}")
           print(f"Title: {book['title']}")
           print(f"Author: {book['author']}")
           print(f"Availability: {'Available' if book['availability'] else 'Not Available'}")
           print()
           print("==================================")
           return
    else:
        print("Book not found in the library.")
def borrow_book():
    book_id = input("Enter book ID    ")
    for book in books:
        if book["id"] == book_id:
           if book["availability"]:
              book["availability"] = False
              print(f"Book '{book['title']}' has been borrowed.")
              return
           else:
              print("Book is not available for borrowing.")
              return
    else:
        print("Book not found in the library.")
        return
def return_book():
    book_id = input("Enter book ID: ")
    for book in books:
        if book["id"] == book_id:
           if not book["availability"]:
              book["availability"] = True
              print(f"Book '{book['book_title']}' has been returned.")
              return
           else:
              print("Book is already available in the library.")
              return
    else:
        print("Book not found in the library.")
        return
while True:
    print("\n======Library Book Management System Menu======:")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")
    choice = input("Enter your choice: ")
    if choice == "1":
        add_book()
    elif choice == "2":
        display_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        borrow_book()
    elif choice == "5":
        return_book()
    elif choice == "6":
        print("Exit")
        break
    else:
        print("Invalid choice. Please try again.")
```
