```python
#student marks
sub1 = int(input("Enter marks of Subject 1: "))
sub2 = int(input("Enter marks of Subject 2: "))
sub3 = int(input("Enter marks of Subject 3: "))

total = sub1 + sub2 + sub3
stu_per = (total / 300) * 100 # Assuming max marks for each subject is 100

print("Total Marks:", total)
print("Percentage:", stu_per)

if stu_per >= 90 and stu_per <= 100:
    print("Grade A+")

elif stu_per >= 80 and stu_per < 90:
    print("Grade A")

elif stu_per >= 70 and stu_per < 80:
    print("Grade B")

elif stu_per >= 60 and stu_per < 70:
    print("Grade C")

elif stu_per >= 50 and stu_per < 60:
    print("Grade D")
elif stu_per >= 30 and stu_per < 50:
    print("Pass")
else:
  print("Fail")
```
