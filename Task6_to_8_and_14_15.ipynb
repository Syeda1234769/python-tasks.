```python
  #Task6
total_bill = 0
total_products=int(input("enter the total_products  "))

for i in range(1, total_products+1):
  product_name=input(f"enter product_name {i}: ")
  product_price=int(input(f"enter product_price of {product_name}  "))
  product_quantity=int(input(f"enter product_quantity of {product_name}   "))

  subtotal = product_price*product_quantity
  total_bill = total_bill + subtotal
if total_bill >= 5000:
    discount = total_bill * 10 / 100
    final_bill = total_bill+ discount
else:
     discount=0
     final_bill =total_bill
print("============/nGrocery Shoping Bill/n============")
print(f"total bill:{total_bill}")
print(f"discount :{discount}")
print(f"final bill:{final_bill}")
```

```python
#task 14 Message Analyzer
message = input("enter your message     ")

total_character = len(message)

total_words = len(message.split())

vowels = "aeiouAEIOU"
vowels_count = 0
for char in message:
 if char in vowels:
    vowels_count +=1
    total_spaces = message.count(" ")
    upper_case = message.upper()

if "python" in message:
  python_exist = "yes"
else:
    python_exist = "no"
print("======Message Analyzer=======")
print("message:", message)
print("total characters:", total_character)
print("total words:", total_words)
print("total vowels:", vowels)
print("total spaces:" , total_spaces)
print("uppercase formatting:", upper_case)
print("python exist:", python_exist)
```

```python
# task 15 Contact Book
contacts= {"Azka":"03001234567","Saba":"03111234567"}
while True:
  print("\n1.add contact")
  print("2.search contacts")
  print("3.display all contacts")
  print("4.exit")
  choice = input("enter your choice:   ")

  if choice == "1":
   name = input("enter contact name   ")
   phone_number = int(input("enter phone number    "))

   if name in contacts:
     print("contact already exist")
   else:
       contacts[name] = phone_number
       print("Contact added successfully")


  elif choice == "2":
      name = input("search name")
      if name in contacts:
       print(f"phone number of {name} is {contacts[name]}")
      else:
       print("contact not found")


  elif choice == "3":
    if contacts:
      for name, phone in contacts.items():
        print(f"{name}:{phone}")
    else:
      print("no contacts found")

  elif choice == "4":
    print("exit program")
    break
else:
  print("Invalid choice")
```
