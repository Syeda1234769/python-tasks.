```python
#task 14 Message Analyzer
message = input("enter your message     ")

total_character = len(message)

total_words = len(message.split())

vowels = "aeiouAEIOU"
vowels_count = 0
for char in message:
 if char in vowels:
    vowels_count +=1
    total_spaces = message.count(" ")
    upper_case = message.upper()

if "python" in message:
  python_exist = "yes"
else:
    python_exist = "no"
print("======Message Analyzer=======")
print("message:", message)
print("total characters:", total_character)
print("total words:", total_words)
print("total vowels:", vowels)
print("total spaces:" , total_spaces)
print("uppercase formatting:", upper_case)
print("python exist:", python_exist)
```

```python

# Task:16 Restaurant Ordering System
menu= {
    "Burger":500,
    "Pizza":1200,
    "Pasta":800,
    "Drink":200
    }
orders=[]
print("=====Restaurant Menu=====")
for key,value in menu.items():
  print(f'{key} : {value}')
while True:
  item_name=input("enter your item name:     ").title()
  if item_name in menu:
   quantity=int(input("enter your quantity:     "))
   price = menu[item_name]
   sub_total = price * quantity
   orders.append({
       "item":item_name,
       "quantity": quantity,
       "subtotal": sub_total
   })
  else:
    print("item is not available")
  choice=input("you want to add more items:yes\no:")
  if choice == "no":
    break
  total = 0
  for order in orders:
   total = total + order["subtotal"]
   tax = (total*13)/100
if total>= 3000:
  discount = total * 10
else:
  discount = 0
grand_total = total + tax - discount
def receipt():
  print("\n=====Receipt=====")
  for order in orders:
    print(f'{order["item"]} x {order["quantity"]} = {order["subtotal"]}')

print("Total:", total)
print("Tax:", tax)
print("Discount:", discount)
print("Grand Total:", grand_total)
```

```python
# Task 17: Clinic Appointment Manager
patients = []

def add_patient():
  patient_id = int(input("Enter your id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient already exists")
      return
  patient_name = input("Enter patient name: ")
  patient_age = int(input("Enter patient age: "))
  problem = input("Enter patient problem: ")
  time = int(input("Enter appointment time: "))
  emergency = input("Is emergency pateint? yes/no: ")

  patients.append({
    "id": patient_id,
    "name": patient_name,
    "age": patient_age,
    "problem": problem,
    "time": time,
    "emergency": emergency
  })

def search_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
      if patient["id"] == patient_id:
         print("Patient found")
         print(f'Patient name: {patient["name"]}')
         print(f'Patient age: {patient["age"]}')
         print(f'Patient problem: {patient["problem"]}')
         print(f'Patient time: {patient["time"]}')
         print(f'Patient emergency: {patient["emergency"]}')
         return
      else:
       print("Patient not found")

def display_appointment():
  print("========== Emergency pateint ==========")
  for patient in patients:
    if patient["emergency"] == "yes":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

  print("========== normal pateint ==========")
  for patient in patients:
    if patient["emergency"] == "no":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

def delete_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      patients.remove(patient)
      print("Patient deleted")
      return
    else:
      print("Patient not found")

def count_patient():
  emergency_count = 0
  normal_count = 0
  for patient in patients:
    if patient["emergency"] == "yes":
      emergency_count += 1
    else:
      normal_count += 1
  print("Emergency patients:", emergency_count)
  print("Normal patients:", normal_count)
  print("Total patients:", len(patients))

while True:
  print("1. Add patient")
  print("2. Search patient")
  print("3. Display appointment")
  print("4. Delete patient")
  print("5. Count patient")
  print("6. Exit")

  choice = int(input("Enter your choice: "))

  if choice == 1:
    add_patient()
  elif choice == 2:
    search_patient()
  elif choice == 3:
    display_appointment()
  elif choice == 4:
    delete_patient()
  elif choice == 5:
    count_patient()
  elif choice == 6:
    print("exit")
    break
  else:
    print("Invalid choice")
```
